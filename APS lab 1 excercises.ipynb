```python
import random # random module is used for generating random outcomes

heads=0
tails=0

for i in range(100):
    toss= random.choice(["Heads","Tails"])

    if toss=="Heads":
        heads+=1
    else:
        tails+=1


#Experimental probablity
probablity_heads = heads/100
probablity_tails = tails/100

print('No of heads:',heads)
print('No of tails:', tails)
print('Experimental probablity of heads:', probablity_heads)
print('Experimental probablity of tails:', probablity_tails)
```

```python
import random

trails=[1000] #setting the no. of trails to check effect of no. of trails

for n in trails:
    heads=0
    tails=0

    for i in range(n):
        toss = random.choice(["Heads","Tails"])

        if toss=="Heads":
            heads += 1
        else:
            tails +=1

probablity_heads = heads/n
probablity_tails = tails/n

print("no of trails:",n)
print("Heads:", heads)
print("Tails:", tails)
print(" Experimental probablity of heads:", probablity_heads)
print("Experimental probablity of tails", probablity_tails)
print('---------------------------------')
```

```python
import random


n=6000 #setting the no. of trails as 6000

counts=[0,0,0,0,0,0] #setting count of each outcome to 0

for i in range(n):
    roll= random.randint(1,6)  # randomly generates a no between 1 and 6, simulates roll of die
    counts[roll-1] += 1  #increases the count for whichever no was rolled

for i in range(6):
    probablity= counts[i]/n

    print("Outcome:", i+1)
    print("Number of trails:", counts[i])
    print("Experimental probablity:", probablity)
```

```python
import random

n = 10000

rolls = [random.randint(1,6) for i in range(n)]

A= sum(1 for roll in rolls if roll ==1)
B= sum(1 for roll in rolls if roll ==6)

A_intersection_B = 0

A_union_B = A + B

P_A = A / n                     #calculate probablities
P_B = B / n
P_A_union_B = A_union_B / n
P_A_plus_P_B = P_A + P_B

print("No of rolls:",n)          #display results
print("A(Getting 1):", A)
print("B(Getting 6):",B)

print("Are A and B disjoint?", A_intersection_B == 0)

print("\n P(A)=", P_A)
print("P(B)", P_B)
print(" P( A U B)=", P_A_union_B)
print("P(A)+P(B)=", P_A_plus_P_B)


if abs(P_A_union_B - P_A_plus_P_B) < 1e-10:
    print("\n Addition rule verified: P(A U B)= P(A)+P(B)")  #verify addition rule
else:
    print("\n Addition rule verified approximately")
```

```python
import random

n=10000

A_count = 0
B_count = 0
A_union_B_count = 0
A_intersection_B_count = 0

for i in range (n):
    roll = random.randint(1,6)

    A = roll in [2,4,6]
    B = roll in [4,5,6]

    if A:
        A_count +=1

    if B:
        B_count +=1

    if A or B:
        A_union_B_count += 1

    if A and B:
        A_intersection_B_count +=1

P_A = A_count/n
P_B = B_count/n
P_union = A_union_B_count/n
P_intersection = A_intersection_B_count/n


print("P(A)=", P_A)
print("P(B)=", P_B)
print("P(A intersection B)=", P_intersection)
print("P(A U B)=", P_union)

print("\n Checking addition rule:")

print("P(A) + P(B)- P(A intersection B)=", P_A+P_B-P_intersection)
print("P(A U B)=", P_union)
```
