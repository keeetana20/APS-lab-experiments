```python
import random
import matplotlib.pyplot as plt


trials = 365                  # Number of simulations for each group size

people_counts = [ 10, 15, 40, 76, 250, 320, 50]             # Different numbers of people

probabilities = []

# Run the simulation for each group size
for n in people_counts:
    matching_trials = 0

    for _ in range(trials):

        
        birthdays = []                  # Generate birthdays

        for _ in range(n):
            birthday = random.randint(1, 365)
            birthdays.append(birthday)

       
        if len(birthdays) != len(set(birthdays)):        # Check for duplicate birthdays
            matching_trials += 1

   
    probability = matching_trials / trials                # Calculate experimental probability
    probabilities.append(probability)

    print("People:", n, "Probability:", probability)


# Plot the results
plt.plot(people_counts, probabilities, marker='o')

plt.xlabel("Number of People")
plt.ylabel("Probability of Shared Birthday")
plt.title("Birthday Problem Simulation")

plt.grid(True)
plt.show()
```
